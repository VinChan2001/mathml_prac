```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array ([[-1,3],[3,2]],dtype=float) # Coefficients of the system of equations
b = np.array ([7,1],dtype=float) # Constants on the right-hand side

print("Coefficient matrix A:", a)
print("Constants vector b:", b)
```

```python
# assuming the system of equations is: -1*x1 + 3*x2 = 7 and 3*x1 + 2*x2 = 1
#we can solve the system of equations using numpy's linear algebra solver
solution = np.linalg.solve(a, b)
print("Solution (x1, x2):", solution)
```

```python
# we can also find the determinant of the coefficient matrix to check if the system has a unique solution
det = np.linalg.det(a)
print("Determinant of A:", det)
```

```python
A_system = np.hstack((a, b.reshape(2, 1))) # Augmented matrix [A|b]
print("Augmented matrix [A|b]:\n", A_system)
```

```python
A_2 = np.array([[-1,3], [3, -9]], dtype=float) # Coefficient matrix with the second column replaced by b

b_2= np.array([7, 1], dtype=float) # Constants vector

d_2 = np.linalg.det(A_2)

print("Determinant of A_2:", np.round(d_2))
```

```python
try:
    solution_2 = np.linalg.solve(A_2, b_2)
    print("Solution for A_2 (x1, x2):", solution_2)
except np.linalg.LinAlgError:
    print("The system of equations has no unique solution.")
```

```python
A_2_system = np.hstack((A_2, b_2.reshape(2, 1))) # Augmented matrix [A_2|b_2]   
A_2_system
```
