```python
import numpy as np
```

```python
A = np.array([[4,-3,1],[2,1,3],[-1,2,-5]],dtype=float)

b = np.array([-10,0, 17], dtype=float)
```

```python
A.shape
```

```python
b.shape
```

```python
x = np.linalg.solve(A, b)
print("Solution for (x1, x2, x3):", x)
```

```python
d = np.linalg.det(A)
print("Determinant of A:", np.round(d))
```

```python
A_2 = np.array([[1,1,1],[0,1,-3],[2,1,5]], dtype=float)

b_2 = np.array([2, 1, 0], dtype=float)

print(np.linalg.solve(A_2, b_2))
```

```python
d_2 = np.linalg.det(A_2)

print("Determinant of A_2:", np.round(d_2))
```
