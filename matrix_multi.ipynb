```python
import numpy as np
```

```python
A = np.array([[4, 9, 9], [9, 1, 6], [9, 2, 3]])
print("Matrix A (3 by 3):\n", A)

B = np.array([[2, 2], [5, 7], [4, 4]])
print("Matrix B (3 by 2):\n", B)
```

```python
# we multiply two matrices using A and B 

np.matmul(A, B) # this is the same as A @ B
```

```python
A @ B # this is the same as np.matmul(A, B) we use @ for dot product and matrix multiplication in numpy
```

```python
# the matmul is only valid when the number of columns in the first matrix (A) matches the number of rows in the second matrix (B).
# In this case, A has 3 columns and B has 3 rows, so the multiplication is valid. The resulting matrix will have dimensions equal 
# to the number of rows of the first matrix (A) and the number of columns of the second matrix (B), which is 3 by 2.

try:
    B@A
except ValueError as e:
    print("Error:", e)
```
