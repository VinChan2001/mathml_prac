```python
import numpy as np
```

```python
parameters = {'W': np.array([[-0.00607548, -0.00126136]]), 'b': np.array([[0.]])}
```

```python
def forward_propagation(X, parameters):
    W = parameters['W']
    b = parameters['b']
    Z = W@X + b
    return Z

def compute_cost(Z,Y):
    m = Y.shape[1] # number of examples
    cost = np.sum((Z-Y)**2) / (2*m)
    return cost

def linear_regression_model(X, Y, number_of_iterations, parameters):

    parameters = parameters

    for i in range(number_of_iterations):

        Z = forward_propagation(X, parameters)
        cost = compute_cost(Z, Y)

        if i % 100 == 0:
            print(f"Cost after iteration {i}: {cost}")
    return parameters
```

```python
#create a simple dataset x1 , x2 and y

X = np.array([[-0.81677709, -0.43287823, -0.94850816], [-0.40838854, -0.21643911, -0.47425408]])
Y = np.array([[0.], [0.], [1.]])

parameters = linear_regression_model(X, Y, 1000, parameters)
```

```python
def predict(X, parameters):
    Z = forward_propagation(X, parameters)
    return Z
```

```python
y_hat = predict(X, parameters)
print("Predicted values:", y_hat)
print("Actual values:", Y)
```
